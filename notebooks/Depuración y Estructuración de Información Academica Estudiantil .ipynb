Construimos una variable que permitiera identificar un estudiante cuantas materias ha perdido esto podria ser un indicador clave de deserción

```python
# Crear la columna con el total de bajo rendimiento por estudiante
total_bajo_rendimiento = df_filtrado.groupby('DOCUMENTO')['BAJO_RENDIMIENTO_REA'].sum().reset_index()
total_bajo_rendimiento
```

```python
# Renombramos la nueva columna
total_bajo_rendimiento = total_bajo_rendimiento.rename(columns={'BAJO_RENDIMIENTO_REA': 'TOTAL_BAJO_RENDIMIENTO_REA'})
total_bajo_rendimiento
```

```python
# Unimos al DataFrame original (si quieres mantenerla allí)
df_filtrado = df_filtrado.merge(total_bajo_rendimiento, on='DOCUMENTO', how='left')
df_filtrado
```

```python
df_promedios = df_filtrado.groupby(['DOCUMENTO', 'AREA_FORMACION'])['REA'].mean().reset_index()
df_pivot = df_promedios.pivot(index='DOCUMENTO', columns='AREA_FORMACION', values='REA')
df_pivot = df_pivot.reset_index()
df_resumen = df_pivot.merge(total_bajo_rendimiento, on='DOCUMENTO', how='left')
df_pivot
```

Hallando inconsistencias, no puede suceder que un graduado no vea todos los espacios académicos o que no tenga notas en alguna area de formación.

```python
areas_clave = [
    'CIENCIAS BÁSICAS',
    'CIENCIAS BÁSICAS DE INGENIERÍA',
    'INGENIERÍA APLICADA',
    'INSTITUCIONAL'
]
inconsistentes = df_resumen_fusagasuga[
    (df_resumen_fusagasuga['SITUACION_ACTUAL'] == 'GRADUADO') &
    (df_resumen_fusagasuga[areas_clave].isna().any(axis=1))
]
inconsistentes
```

```python
# Paso 1: Definir las áreas de formación clave
areas_clave = ['CIENCIAS BÁSICAS', 'CIENCIAS BÁSICAS DE INGENIERÍA', 'INGENIERÍA APLICADA', 'INSTITUCIONAL']

# Paso 2: Crear máscara booleana para inconsistencias
mask_inconsistentes = (
    (df_resumen_fusagasuga['SITUACION_ACTUAL'].str.strip().str.upper() == 'GRADUADO') &
    df_resumen_fusagasuga[areas_clave].isna().any(axis=1)
)

# Paso 3: Crear DataFrame limpio (excluyendo los inconsistentes)
df_resumen_limpio = df_resumen_fusagasuga[~mask_inconsistentes].reset_index(drop=True)
```

```python
df_resumen_limpio
```

```python
df_resumen_limpio["SITUACION_ACTUAL"].value_counts()
```

```python
# Mapeo personalizado
mapa_desercion = {
    'ACTIVO': 0,
    'GRADUADO': 0,
    'REVISION CARPETA GRADO': 0,

    'INACTIVO': 1,
    'EXCLUIDO NO RENOVACION DE MATRICULA': 1,
    'EXCLUIDO BAJO RENDIMIENTO': 1,
    'EXCLUIDO CANCELACION SEMESTRE': 1,
    'EXCLUIDO TRANSFERENCIA INTERNA': 1,
    'EXCLUIDO TRANSFERENCIA INTERNA SA': 1,
    'EXCLUIDO ASIMILACION DE PENSUM': 1,
    'SUSPENDIDO': 1,
    'SUSPENSION UN PERIODO ACADEMICO': 1
}

# Nueva variable binaria
df_resumen_limpio['DESERTA'] = df_resumen_limpio['SITUACION_ACTUAL'].map(mapa_desercion)
```

# Conlusiones
 - Se logró consolidar un conjunto de datos estructurado por estudiante, en el cual cada fila representa un registro único con su promedio de desempeño (REA) en cada una de las áreas de formación: Ciencias Básicas, Ciencias Básicas de Ingeniería, Ingeniería Aplicada e Institucional.

 - Se identificaron inconsistencias académicas en estudiantes cuya situación académica era “GRADUADO” pero que no tenían registro de notas en áreas críticas, lo cual permitió depurar la base y conservar únicamente los casos coherentes con el proceso formativo.

 - Se construyó una nueva variable binaria (DESERTA) a partir de la situación académica (SITUACION_ACTUAL), que permitirá abordar tareas de clasificación supervisada para la predicción del riesgo de deserción estudiantil.

 - La variable TOTAL_BAJO_RENDIMIENTO_REA permitió capturar el número de asignaturas en las que el estudiante ha obtenido un rendimiento bajo (nota menor a 3.0), generando un importante indicador de alerta temprana.

 - El conjunto final fue filtrado para incluir únicamente estudiantes de la sede Fusagasugá, permitiendo un análisis focalizado y alineado con la realidad local del programa académico.
